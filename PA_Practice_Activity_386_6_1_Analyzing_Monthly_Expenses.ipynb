```python
import numpy as np
import matplotlib.pyplot as plt

#Assigning x and y co-ordinates
Exp_Category = ['Groceries', 'Utilities', 'Transportation', 'Dining Out','Entertainment']
Amount_Spent = [500, 300, 200, 400, 250]

#Assigning colors for the bars
colors = ["blue", "yellow",  "lightgreen", "red", "pink"]

# Plotting hbar chart
plt.barh(Exp_Category,Amount_Spent, color= colors, height = 0.7)

# Displaying the Labels
plt.xlabel('Amount Spent',fontsize=14)
plt.ylabel('Expense Categories',fontsize=14)

# Displaying the title
plt.title('Monthly Expenses Distribution',fontsize=20)

plt.show()
```
