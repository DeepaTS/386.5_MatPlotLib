# **Section One**

1. Read Data from the 'woocommerce-product-export.csv' file using the python library.
2. Show a concise summary of the columns using the info() method.
3. Show a summary of statistics pertaining to the columns.
4. Print the first five rows by default.
5. Print the last five rows by default.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python

# 1. Read the CSV file
df = pd.read_csv("woocommerce-product-export.csv")
print(df)

#2.Concise summary of the columns
print("DataFrame Info:")
print("----------------")
df.info()

#3. summary of statistics
print("\nSummary of statistics:")
print("------------------------")
print(df.describe(include="all"))

#4.First 5 rows
print(df.head())

#5.Last 5 rows
print(df.tail())
```

### **6. Print  the "total_profit" and “month_number” columns only.**

```python
# Print only the 'month_number' and 'total_profit' columns
print(df[['month_number', 'total_profit']])
```

### **7. Read the total profit of all months and show it using the Bar plot.**

  The generated Bar plot must include the following properties:

  X label name = Month number.

  Y label name = Total profit.

  Title = Company profit per month.


```python

# Create a bar plot
plt.bar(df['month_number'], df['total_profit'], color='blue')

# Add labels and title
plt.xlabel("Month number")
plt.ylabel("Total profit")
plt.title("Company profit per month")

plt.yticks(range(0,600000, 100000))

# Display the plot
plt.grid(linestyle = "dashed")
plt.show()
```

### **8. Read the total profit of all of the months, and show the line plot with    the following style properties:**


  The generated line plot must include the following Style properties:

  The line style should be dotted and the line color should be red.

  Show the legend at the lower-right location.

  X label name = Month number

  Y label name = Total profit.

  Title = Company profit per month.

  Add a circle marker; the line marker color = black.

  The line width should be 3.






```python
# Create the line plot
plt.plot(
    df["month_number"],
    df["total_profit"],
    linestyle=":",          # Dotted line
    color="red",            # Line color
    linewidth=3,            # Line width
    marker="o",             # Circle marker
    markerfacecolor="black",# Marker color
    label="Profit data of last year"
   )

# Labels and title
plt.xlabel("Month number")
plt.ylabel("Total profit")
plt.title("Company profit per month")

# Show every month on the x-axis
plt.xticks(range(1, 13))
plt.yticks(range(0,600000, 100000))

# Show legend at lower-right
plt.legend(loc="lower right")

# Display the plot
# plt.grid(linestyle = "dashed")
plt.show()
```

### 9. **Print all of the product sales data and show it using a multi-line plot.**

Display the number of units sold per month for each product using multi-line plots. (i.e., show a separate plotline for each product).


```python
# line Plot sales for each product
plt.plot(df["month_number"], df["facecream"], label="Face Cream Sales Data",marker="o",color="blue")
plt.plot(df["month_number"], df["facewash"], label="Face Wash Sales Data",marker="o",color="Orange")
plt.plot(df["month_number"], df["toothpaste"], label="Toothpaste Sales Data",marker="o",color="green")
plt.plot(df["month_number"], df["bathingsoap"], label="Bathing Soap Sales Data",marker="o",color="red")
plt.plot(df["month_number"], df["shampoo"], label="Shampoo Sales Data",marker="o",color="purple")
plt.plot(df["month_number"], df["moisturizer"], label="Moisturizer Sales Data",marker="o",color="brown")

# Labels and title
plt.xlabel("Month Number")
plt.ylabel("Sales units in number")
plt.title("Sales Data")

# Show every month on the x-axis
plt.xticks(range(1, 13))
plt.yticks(range(0,20000,2000))

# Show legend
plt.legend()

# Display the graph
plt.show()
```

### **10. Read “bathingsoap” sales data for each month and show it using a scatter plot.**
Add a grid to the plot. The grid line style should be “–“.


```python
# Create the Scatter plot
plt.scatter(df["month_number"], df["bathingsoap"], color="blue",label='bathingsoap Sales Data')
# Labels and title
plt.xlabel("Month Number")
plt.ylabel("Number of units Sold")
plt.title("bathingsoap Sales Data")

# Show every month on the x-axis
plt.xticks(range(1, 13))

# Show legend
plt.legend()

# Display the graph
plt.grid(linestyle = "--")
plt.show()
```

# **2.Given Dataset:**
height=[121.9,124.5,129.5,134.6,139.7,147.3,152.4,157.5,162.6]

weight=[19.7,21.3,23.5,25.9,28.5,32.1,35.7,39.6,43.2]

Show the average weight against the average height (use a line chart or any other chart):

The generated **line plot** must include the following Style properties:

The line style should be dash-dot and the line color should be green.

Show the legend at the lower-right location.

X label name = Weight in kg.

Y label name = Height in cm.

Title = Average weight with respect to average height.

Add a circle marker: marker size = 10; circle marker color = green.


```python

height=np.array([121.9,124.5,129.5,134.6,139.7,147.3,152.4,157.5,162.6])

weight=np.array([19.7,21.3,23.5,25.9,28.5,32.1,35.7,39.6,43.2])

# Create line plot
plt.plot(
    weight,
    height,
    linestyle="-.",      # Dash-dot line
    color="green",       # Green line
    linewidth=2,
    marker="o",          # Circle marker
    markersize=10,       # Marker size
    markerfacecolor="green",
    markeredgecolor="green",
    label="Average Height"
)

# Labels and title
plt.xlabel("Weight in kg")
plt.ylabel("Height in cm")
plt.title("Average weight with respect to average height")

# Show legend at lower-right
plt.legend(loc="lower right")

# Display the graph
plt.show()
```
