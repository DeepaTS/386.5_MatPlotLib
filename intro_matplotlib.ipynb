**State-based (single object) Plot**

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
xpoints = np.array([1, 8])
ypoints = np.array([3, 10])

# Styles
styles = {
    'color': 'red',
    'linewidth': 2,
    'linestyle': '--'
}

# State-based (single object) Plot
plt.plot(xpoints, ypoints, **styles)
plt.title('Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()


```

**Object Oriented [Two objects: figure and axe]**
https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplots.html

```python
# Data
xpoints = np.array([1, 8])
ypoints = np.array([3, 10])

# Styles
styles = {
    'color': 'red',
    'linewidth': 2,
    'linestyle': '--'
}

# Object Oriented (Two objects)
fig, ax = plt.subplots()
fig.suptitle('Figure Title')

# Here we mostly work with the axe object:
ax.plot(xpoints, ypoints, **styles)
# ax.title('Line Plot') <- This is not a method of the axe object
ax.set_title('Line Plot Title') # < use this methos to set the title
print(ax.title) # <- title is a attribute of the axe object that return the title value
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
plt.show()
```

**Subplot** https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplot.html

```python
# Data
xpoints = np.array([1, 8])
ypoints = np.array([3, 10])

# Styles
styles = {
    'color': 'red',
    'linewidth': 2,
    'linestyle': '--'
}


plt.subplot(1, 2, 1) # <- index 1
plt.plot(xpoints, ypoints, **styles)
plt.title('Line Plot 1')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')

# We can create a new variable or just selected by the index
plt.subplot(1, 2, 2) # <- index 2
plt.plot(xpoints, ypoints, **styles)
plt.title('Line Plot 2')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')

plt.tight_layout()
plt.show()
```

```python
# Data
xpoints = np.array([1, 8])
ypoints = np.array([3, 10])

# Styles
styles = {
    'color': 'red',
    'linewidth': 2,
    'linestyle': '--'
}

# Create a figure with size: 8x8
plt.figure(figsize=(8, 8))

# Create a 2 by 2 grid and select plot 1
plt.subplot(2, 2, 1) # <- select index 1
plt.plot(xpoints, ypoints, **styles)
plt.title('Line Plot 1')
plt.xlabel('X-axis 1')
plt.ylabel('Y-axis 1')

# Plot 2
plt.subplot(2, 2, 2) # <- select index 2
plt.plot(xpoints, ypoints, **styles)
plt.title('Line Plot 2')
plt.xlabel('X-axis 2')
plt.ylabel('Y-axis 2')


# Plot 3
plt.subplot(2, 2, 3) # <- select index 3
plt.plot(xpoints, ypoints, **styles)
plt.title('Line Plot 3')
plt.xlabel('X-axis 3')
plt.ylabel('Y-axis 3')

# Plot 4
plt.subplot(2, 2, 4) # <- select index 4
plt.plot(xpoints, ypoints, **styles)
plt.title('Line Plot 4')
plt.xlabel('X-axis 4')
plt.ylabel('Y-axis 4')

plt.tight_layout()
plt.show()
```

**Object Oriented (Plot and Axe objects)**

```python
# Data
xpoints = np.array([1, 8])
ypoints = np.array([3, 10])

# Styles
styles = {
    'color': 'red',
    'linewidth': 2,
    'linestyle': '--'
}

# Create a 2x2 grid at once
fig, axes = plt.subplots(2, 2, figsize=(8, 8))

# Unpack the 2D array: axes[row, col]
axes[0, 0].plot(xpoints, ypoints, **styles)
axes[0, 0].set_title('Line Plot 1')

axes[0, 1].plot(xpoints, ypoints, **styles)
axes[0, 1].set_title('Line Plot 2')

axes[1, 0].plot(xpoints, ypoints, **styles)
axes[1, 0].set_title('Line Plot 3')

axes[1, 1].plot(xpoints, ypoints, **styles)
axes[1, 1].set_title('Line Plot 4')

# Set shared labels for all axes easily in a loop
for ax in axes.flat:
    ax.set_xlabel('X-axis')
    ax.set_ylabel('Y-axis')

plt.tight_layout()
plt.show()
```
